# PV Generation Profile using pvlib

This notebook demonstrates how to generate photovoltaic (PV) generation profiles from weather data using the `pvlib` Python library.

## Overview

We calculate solar irradiance on different surface orientations (flat rooftop and façade) using weather data and solar position calculations for Zurich, Switzerland.

## Key Components

### 1. Solar Position Calculation
- **Documentation**: [pvlib.solarposition.get_solarposition](https://pvlib-python.readthedocs.io/en/stable/reference/generated/pvlib.solarposition.get_solarposition.html#examples-using-pvlib-solarposition-get-solarposition)
- **Example**: [Sun path diagrams](https://pvlib-python.readthedocs.io/en/stable/gallery/solar-position/plot_sunpath_diagrams.html#sphx-glr-gallery-solar-position-plot-sunpath-diagrams-py)
- **Location**: Zurich coordinates (latitude/longitude/altitude)

### 2. Total Irradiance Calculation
- **Documentation**: [pvlib.irradiance.get_total_irradiance](https://pvlib-python.readthedocs.io/en/stable/reference/generated/pvlib.irradiance.get_total_irradiance.html#pvlib.irradiance.get_total_irradiance)
- **Model**: Perez (recommended for accurate diffuse irradiance modeling)
- **Output**: `poa_global` (plane-of-array global irradiance)

### 3. Surface Configurations

**Flat Rooftop:**
- For horizontal surfaces (tilt = 0°), Global Horizontal Irradiance (GHI) serves as a good approximation for `poa_global`
- PV output = GHI × conversion efficiency × panel area

**Façade:**
- Requires recalculation using `get_total_irradiance` with appropriate tilt angle (typically 90° for vertical façades)
- Surface azimuth determines orientation (e.g., 180° for south-facing)


# create local time
NOTE: 2016 has 366 days, the data for last day (31st of December) is removed to keep number of days as 365

```python
import pandas as pd
import pytz

# Define the timezone for Switzerland
switzerland_tz = pytz.timezone('Europe/Zurich')

# Create DateTimeIndex for 2016, hourly intervals, in local time at Switzerland
start_date = pd.Timestamp('2016-01-01 00:00:00', tz=switzerland_tz)
end_date = pd.Timestamp('2016-12-30 23:00:00', tz=switzerland_tz)

datetime_index = pd.date_range(start=start_date, end=end_date, freq='H')

# Display the DateTimeIndex
datetime_index
```

## get solar position

```python
import pvlib

# get solar position by the latidude and logitude of Zurich
df_solar_pos = pvlib.solarposition.get_solarposition(datetime_index, 47.372582, 8.532812)

# when apparent_elevation, set it to 0 (at night)
df_solar_pos.loc[df_solar_pos['apparent_elevation']<0,'apparent_elevation'] = 0

df_solar_pos.head(24)
```

# read weather data from CESAR-P

```python
# NOTE: this weather data is in the csv format
path_weather = r'G:\02-03 Projects-Ongoing\BBL\01-01 Datasets\weather_data\\'
df_pv_profile = pd.DataFrame()
df_pv_profile.index = datetime_index

# get extra radiation
extra_rad = pvlib.irradiance.get_extra_radiation(datetime_index)

for cz in range(6):
    df_weather = pd.read_csv(path_weather + f'weather_clim_{cz}.csv')
    df_weather.index = datetime_index

    # get total irradiance - flat
    df_total_irr_90 = pvlib.irradiance.get_total_irradiance(
        surface_tilt = 0,
        surface_azimuth = 180,
        solar_zenith = df_solar_pos['zenith'],
        solar_azimuth = df_solar_pos['azimuth'],
        dni = df_weather['Direct Normal Radiation'],
        ghi = df_weather['Global Horizontal Radiation'],
        dhi = df_weather['Diffuse Horizontal Radiation'],
        dni_extra = extra_rad,
        model = 'perez'
    )
    df_pv_profile[f'flat_{cz}'] = df_total_irr_90['poa_global']

    # get total irradiance - fassade
    df_total_irr_90 = pvlib.irradiance.get_total_irradiance(
        surface_tilt = 90,
        surface_azimuth = 180,
        solar_zenith = df_solar_pos['zenith'],
        solar_azimuth = df_solar_pos['azimuth'],
        dni = df_weather['Direct Normal Radiation'],
        ghi = df_weather['Global Horizontal Radiation'],
        dhi = df_weather['Diffuse Horizontal Radiation'],
        dni_extra = extra_rad,
        model = 'perez'
    )
    df_pv_profile[f'fassade_south_{cz}'] = df_total_irr_90['poa_global']

df_pv_profile
```
